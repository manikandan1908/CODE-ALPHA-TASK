- Pandas

## 💡 Skills

- Sentiment Classification

## 🚀 Outcome
This project analyzes customer reviews and classifies sentiment to understand public opinion.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Review": [
        "This product is amazing",
        "Worst experience ever",
        "It is okay, not bad",
        "Absolutely loved it",
        "Very disappointing",
        "Good quality and nice",
        "Not worth the money",
        "Average product",
        "Excellent performance",
        "Bad quality"
    ]
}

df = pd.DataFrame(data)
df
```

```python
from textblob import TextBlob

def get_sentiment(text):
    analysis = TextBlob(text)
    score = analysis.sentiment.polarity

    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Review"].apply(get_sentiment)
df
```

```python
sns.countplot(x='Sentiment', data=df)
plt.title("Sentiment Distribution")
plt.show()
```

# 📊 Sentiment Analysis – Data Story

## 🔍 Insights
- Most reviews are positive, indicating customer satisfaction.
- Some negative reviews highlight product quality issues.
- Neutral reviews show mixed opinions.

## 📈 Interpretation
- Customers generally like the product.
- Negative feedback can help improve product quality.

## 🎯 Conclusion
Sentiment analysis helps understand customer opinion and improve business decisions.
