```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("titanic.csv")
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival")
plt.show()
```

```python
sns.histplot(df['Age'], kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Passenger Class vs Survival")
plt.show()
```

```python
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title("Fare vs Survival")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df[['Survived','Age','Fare','Pclass']])
plt.show()
```

# 📊 Data Story – Titanic Dataset

## 🔍 Key Insights

- Female passengers had a higher survival rate compared to males.
- First-class passengers were more likely to survive.
- Younger passengers (children) had better survival chances.
- Passengers who paid higher fares had a higher probability of survival.

## 📈 Interpretation

The analysis shows that survival was not random. It was influenced by:
- Gender (priority given to women)
- Socio-economic status (class & fare)
- Age (children safety priority)

## 🎯 Conclusion

From the visualizations, we can conclude that:
- Women and children were prioritized during rescue.
- Wealthier passengers had better access to safety measures.
- Data visualization helps uncover hidden patterns and supports better decision-making.
